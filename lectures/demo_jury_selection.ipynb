## Warm-up on probability ##

```python
# Run the experiment

def count_heads():
    return sum(np.random.choice(make_array('heads', 'tails'), 100) == 'heads')

repetitions = 100000
head_counts = make_array()

for _ in np.arange(repetitions):
    head_counts = np.append(head_counts, count_heads())

results = Table().with_columns('Heads', head_counts)
results
```

```python
# Make a histogram of the results. Highlight 50, 60, and 70.
results.hist(bins=np.arange(20, 80))
plots.scatter(50, 0.005, color='green', s=200)
plots.scatter(60, 0.005, color='yellow', s=200)
plots.scatter(70, 0.005, color='red', s=200)
```

```python
# Get some numerical results:

print('Number of heads at least 50: ' 
 + str(results.where('Heads', are.above_or_equal_to(50)).num_rows))
print('Number of heads at least 60: ' 
 + str(results.where('Heads', are.above_or_equal_to(60)).num_rows))
print('Number of heads at least 70: ' 
 + str(results.where('Heads', are.above_or_equal_to(70)).num_rows))
```
