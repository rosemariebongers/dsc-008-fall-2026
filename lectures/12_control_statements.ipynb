 **Simulating heads in 100 coin tosses**

```python
# Set up our coins
coin = make_array('heads', 'tails')
```

```python
# Let's check that np.rand.choice does what we think
np.random.choice(coin, 100)
```

```python
# Do one trial
sum(np.random.choice(coin, 100) == 'heads')
```

```python
# Simulate one outcome

def num_heads(n):
    return sum(np.random.choice(coin, n) == 'heads')
```

```python
# Decide how many times you want to repeat the experiment
repetitions = 1000
```

```python
# Simulate that many outcomes

outcomes = make_array()

for i in np.arange(repetitions):
    outcomes = np.append(outcomes, num_heads(100))
    # The np.append(list, thing) adds thing to list in the final slot
    # and extends the length by 1.
```

```python
# Make a Table() of the results
# and then draw a histogram of it.

heads = Table().with_columns('Head count', outcomes)
heads.hist('Head count', bins=range(30, 70))
```
