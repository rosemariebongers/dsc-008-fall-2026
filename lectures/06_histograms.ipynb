## Height of a bin ##

### Question: What is the height of the [40, 65) bin?

```python
# Step 1: Calculate the percentage of movies released for this bin
movies_in_40_65_bin = top_movies.where('Age', are.below(65)).where('Age', are.above_or_equal_to(40))
percent = movies_in_40_65_bin.num_rows / top_movies.num_rows
percent
```

```python
# Step 2: Calculate the width of the 40-65 bin
width = 65 - 40
```

```python
# Step 3: Area of rectangle = height * width
#         --> height = percent / width
height = 100*percent / width
height
```
